## Customizing Colorbars

```python
from matplotlib.colors import LinearSegmentedColormap

def grayscale_cmap(cmap):

    cmap = plt.cm.get(cmap)
    colors = cmap(np.arange(cmap.N))

    RGB_weight = [0.299, 0.587, 0.114]
    luminance = np.sqrt(np.dot(colors[:, :3] ** 2, RGB_weight))
    colors[:, :3] = luminance[:, np.newaxis]

    return LinearSegmentColormap.from_list(cmap.name + "_gray", colors, cmap.N)
```

```python
def view_colormap(cmap):
    cmap = plt.cm.get_cmap(cmap)
    colors = cmap(np.arange(cmap.N))

    cmap = grayscale_cmap(cmap)
    grayscale = cmap(np.arange(cmap.N))

    fig, ax = plt.subplots(2, figsize=(6, 2),
    subplots_kw = dict(xticks=[], yticks=[]))

    ax[0].imshow([colors], extent=[0, 10, 0, 1]) 
    ax[1].imshow([grayscale], extent=[0, 10, 0, 1])

    view_colormap("jet")
```

## Example: Effect of Holidays on US Births

```python
births = pd.read_csv(r'C:\Users\Admin\Desktop\github\python-data-science-handbook\Projects\data\births in usa.csv')

quartiles = np.percentile(births['births'], [25, 50, 75])

mu, sig = quartiles[1], 0.74 * (quartiles[2] - quartiles[0])

births = births.query('(births > @mu - 5 * @sig) & (births < @mu + 5 * @sig)')

births['day'] = births['day'].astype(int)

births.index = pd.to_datetime(10000 * births.year + 100 * births.month + births.day, format='%Y%m%d')

births_by_date = births.pivot_table('births', [births.index.month, births.index.day])

fig, ax = plt.subplots(figsize=(12, 4))

births_by_date.plot(ax=ax)
```

```python
def format_func(value, ticknumber):
    N = int(np.round(2* value / np.pi))
    if N == 0:
        return "0"
    elif N == 1:
        return r"$\pi/2$"
    elif N == 2:
        return  r"$\pi$"
    elif N % 2 > 0:
        return r"${0}\pi/2$".format(N)
    else:
        return r"${0}\pi$".format(N // 2)
    
ax.xaxis.set_major_formatter(plt.FuncFormatter(format_func))
fig
```

## Plot Customization By Hand

```python
import matplotlib.pyplot as plt
plt.style.use("classic")
import numpy as np

x = np.random.randn(1000)
plt.hist(x)
```

```python
ax = plt.axes(axisbg="#E6E6E6")
ax.set_axisbelow(True)

plt.grid(color="w", linestyle="solid")

for spine in ax.spines.values():
    spine.set_visible(False)

ax.xaxis.tick_bottom()
ax.yaxis.tick_left()

ax.tick_params(color="gray", direction="out")
for tick in ax.get_xticklabels():
    tick.set_color("gray")
for tick in ax.get_yticklabel():
    tick.set_color("gray")

ax.hist(x, edegecolor="#E6E6E6", color="#EE6666")
```
