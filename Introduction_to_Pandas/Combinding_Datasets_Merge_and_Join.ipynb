## Combinding Datasets: Merge and Join

```python
import pandas as pd
import numpy as np
```

### One-to-one joins

```python
df1 = pd.DataFrame({"employee": ["Bob", "Jake", "Lisa", "Sue"],
                    "group": ["Accounting", "Engineering", "Engineering", "HR"]})

df2 = pd.DataFrame({"employee":["Lisa", "Bob", "Jake", "Sue"],
                    "hire_date": [2004, 2008, 2012, 2014]})

print(df1); print(df2)
```

```python
# combinding information into one DataFrame
df3 = pd.merge(left=df1, right=df2)
df3
```

### Many-to-one joins

```python
df4 = pd.DataFrame({"group":["Accounting", "Engineering", "HR"],
                    "supervisor": ["Carly", "Guido", "Steve"]})

print(df3); print(df4)
```

```python
print(pd.merge(df3, df4))
```

### Many-to-many joins

```python
df5 = pd.DataFrame({"group": ["Accounting", "Accounting", "Enginnering", "Engineering", "HR", "HR"],
                    "skills": ["math", "Spreadsheets", "coding", "linux", "Spreadsheets", "organization"]})


print(df1); print(df5)
```

```python
print(pd.merge(df1, df5))
```

## Specification of the Merge Key

### **The on keyword**

```python
print(df1); print(df2)
```

```python
print(pd.merge(df1, df2, on="employee"))
```

### **The left_on and right_on keywords**

```python
df3 = pd.DataFrame({"name":["Bob", "Jake", "Lisa", "Sue"],
                   "salary": [70000, 80000, 120000, 90000]})

print(df1); print(df3)
```

```python
print(pd.merge(df1, df3, left_on="employee", right_on="name"))
```

```python
pd.merge(df1, df3, left_on="employee", right_on="name").drop("name", axis=1)
```

### **The left_index and right_index keywords**

```python
df1a = df1.set_index("employee")
df2a = df2.set_index("employee")
print(df1a); print(df2a)
```

```python
print(df1a); print(df2a)
```

```python
print(pd.merge(df1a, df2a, left_index=True, right_index=True))
```

```python
print(df1a); print(df2a); print(df1a.join(df2a))
```

```python
print(df1a); print(df3)
print(pd.merge(df1a, df3, left_index=True, right_on="name"))
```

## Specifying Set Arithmetic for Joins

```python
df6 = pd.DataFrame({'name': ['Peter', 'Paul', 'Mary'],
                            'food': ['fish', 'beans', 'bread']},
                           columns=['name', 'food'])
df7 = pd.DataFrame({'name': ['Mary', 'Joseph'],
                            'drink': ['wine', 'beer']},
                           columns=['name', 'drink'])
print(df6); print(df7); print(pd.merge(df6, df7))
```

```python
pd.merge(df6, df7, how='inner')
```

```python
print(df6); print(df7); print(pd.merge(df6, df7, how="left"))
```

## Overlapping Column Names: The suffixes Keyword

```python
df8 = pd.DataFrame({"name": ["Bob", "Jake", "Lisa", "Sue"],
                    "rank": [1, 2, 3, 4]})
```

```python
df9 = pd.DataFrame({"name": ["Bob", "Jake", "Lisa", "Sue"],
                    "rank": [3, 1, 4, 2]})
```

```python
print(df8); print(df9); print(pd.merge(df8, df9, on="name"))
```

```python
print(df8); print(df9); print(pd.merge(df8, df9, on="name", suffixes=["_L", "_R"]))
```

## Example: US States Data

```python
pop_path = r"C:\Users\Admin\Desktop\github\python-data-science-handbook\data\state-population.csv"
areas_path = r"C:\Users\Admin\Desktop\github\python-data-science-handbook\data\state-areas.csv"
abbrevs_path = r"C:\Users\Admin\Desktop\github\python-data-science-handbook\data\state-abbrevs.csv"
```

```python
pop = pd.read_csv(pop_path)
areas = pd.read_csv(areas_path)
abbrevs = pd.read_csv(abbrevs_path)
```

```python
print(pop.head()); print(areas.head()); print(abbrevs.head())
```

```python
merged = pd.merge(pop, abbrevs, how="outer", left_on = "state/region", right_on="abbreviation")
merged = merged.drop("abbreviation", axis=1)
merged.head()
```

```python
merged.isnull().any()
```

```python
merged[merged["population"].isnull()].head()
```

```python
merged.loc[merged["state"].isnull(), "state/region"].unique()
```

```python
merged.loc[merged["state/region"] == "PR", "state"] = "Puerto Rico"
merged.loc[merged["state/region"] == "USA", "state"] = "United States"
merged.isnull().any()
```

```python
final = pd.merge(merged, areas, on="state", how="left")
final.head()
```

```python
final.isnull().any()
```

```python
final["state"][final["area (sq. mi)"].isnull()].unique()
```

```python
final.dropna(inplace=True)
final.head()
```

```python
data2010 = final.query("year == 2010 & ages == 'total'")
data2010.head()
```

```python
data2010.columns
```

```python
data2010.set_index("state", inplace=True)
density = data2010["population"]/data2010["area (sq. mi)"]
```

```python
density.sort_values(ascending=False, inplace=True)
density.head()
```
