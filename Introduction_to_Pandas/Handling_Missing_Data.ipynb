# NaN: Missing numerical data

```python
vals2 = np.array([1, np.nan, 3, 4])
vals2.dtype
```

```python
1 + np.nan
```

```python
0 * np.nan
```

```python
vals2.sum(), vals2.min(), vals2.max()
```

```python
np.nansum(vals2), np.nanmin(vals2), np.nanmax(vals2)
```

Nan only works with floating-point numbers so floats.
Any math involing NaN becomes Nan
ignoring NaN function --> they skip the missing value
NaN only exists for floating number so not for strings, integers etc.

### NaN and None in Pandas

# Operating on Null Values

### Detecting null values

```python
data = pd.Series([1, np.nan, 'hello', None])
data.isnull()
```

```python
data[data.notnull()]
```

isnull() → tells you where data is missing
notnull() → tells you where data exists
They return True/False for every element.

```python
# Dropping null values
```

```python
data.dropna()
```

dropna() removes rows or columns that contain missing values (NaN / None)

```python
data
```

```python
data.dropna(inplace=True)
```

```python
data
```

```python
data = pd.Series([10, 11, None, 13])
print(data)
```

```python
df = pd.DataFrame([[1, np.nan, 2],
                   [2,    3,   5],
                   [np.nan, 4, 6]])
df
```

```python
df.dropna()
```

```python
df
```

```python
df.dropna(axis="columns")
```

```python
df.dropna(axis=0)
```

### **Filling null values**

```python
data = pd.Series([1, np.nan, 2, None, 3], index=list("abcde"))
data
```

```python
# filling NaN values with single values like Zeros.
data.fillna(0)
```

```python
# forward-fill 
data.fillna(method="ffill")
```

```python
# back-fill
data.fillna(method="bfill")
```
