## Introduction to Pandas Objects

```python
import numpy as np
import pandas as pd
```

### **The Pandas Series Object**

```python
data = pd.Series([0.25, 0,5, 0.75, 1.0])
data
```

```python
# accessing the values
data.values
```

```python
# the index
data.index
```

```python
# accessing the data with square brackets
data[1]
data[1:3]
```

### **Series as generalized NumPy array**

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=["a","b","c","d"])
data
```

```python
# accessing the item
data["c"]
```

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=[2, 5, 3, 7])
data
```

```python
# accessing the item with the new assigned index
data[3]
```

### **Series as specialized dictionary**

```python
population_dict = {"California": 38332521,
                   "Texas": 26448193,
                   "New York": 19651127,
                   "Florida": 19552860,
                   "Illinois": 12882135}
population = pd.Series(population_dict)
population
```

```python
# index drawn from the sorted keys
population["California"]
```

```python
# Series also supports array-style operations like slicing
population["California":"Florida"]
```

### **Constructing Series objects**

```python
pd.Series([2, 4, 6])
```

```python
pd.Series(5, index=[100, 200, 500])
```

```python
pd.Series({2:"a", 1:"b", 3:"c"})
```

```python
# index can be set if needed
pd.Series({2:"a", 1:"b", 3:"c"}, index=[3, 2])
```

## The Pandas DataFrame Object

```python
area_dict = {"California": 423967, "Texas": 695662, "New York": 141297, "Florida": 170312, "Illinois": 149995}

area = pd.Series(area_dict)
area
```

```python
states = pd.DataFrame({"population": population, "area": area})
states
```

### **Dataframe as specializd dictionary**

```python
states["area"]
```

### Constructing Dataframe objects

```python
pd.DataFrame(population, columns=["population"])
```

```python
# converting lists to dicts
data = [{"a": i, "b": 2 * i}
        for i in range(3)]
pd.DataFrame(data)
```

```python
pd.DataFrame([{"a": 1, "b": 2}, {"b":3, "c": 4}])
```

```python
# converting a dictionary to a Series of Objects
pd.DataFrame({"population": population, 
              "area": area})
```

```python
# converting a two dimensional Numpy to a Series of Objects
pd.DataFrame(np.random.rand(3, 2),
             columns=["foo", "bar"],
             index = ["a", "b", "c"])
```

```python
# converting a Numpy Array to a Series of Objects
A = np.zeros(3, dtype=[("A", "i8"), ("B", "f8")])
A
```

## The Pandas Index Object

```python
ind = pd.Index([2, 3, 5, 7, 11])
ind
```

### **Index as immutable array**

```python
ind[1]
```

```python
ind[::2]
```

```python
print(ind.size, ind.shape, ind.ndim, ind.dtype)
```

### **Index as ordered set**

```python
indA = pd.Index([1, 3, 5, 7, 9])
indB = pd.Index([2, 3, 5, 7, 11])
indA & indB # intersection
```

```python
indA
```

```python
indB
```

```python
indA & indB
```

## Data Indexing and Selection

### **Series as dictionary**

```python
import pandas as pd

data = pd.Series([0.25, 0.5, 0.75, 1.0],
index=["a","b","c","d"])

data
```

```python
"a" in data
```

```python
data.keys()
```

```python
list(data.items())
```

```python
# extending the data
data["e"] = 1.25
data
```

### **Series as one dimensional array**

```python
# slicing by specific index
data["a":"c"]
```

```python
# masking 
data[(data > 0.3) & (data < 0.8)]
```

```python
# fancy indexing
data[["a", "e"]]
```

### **Indexers: Ioc, iloc and ix**

```python
data = pd.Series(["a", "b", "c"], index=[1, 3, 5])
data
```

```python
data[1]
```

```python
data[1:3]
```

```python
data = [
    {"name": "Aditi", "age": 15, "weight": 55}, 
    {"name": "Sneha", "age": 12, "weight": 45},
    {"name": "Yuri", "age": 22, "weight": 66}
    ]
df = pd.DataFrame(data)
df
```

```python
df.iloc[:2, :2]
```

```python
df.loc[0:2, ["name", "weight"]]
```
