## Hierarchical Indexing

```python
import pandas as pd
import numpy as np
```

```python
index = [("California", 2000), ("California", 2010),
         ("New York", 2000), ("New York", 2010), 
         ("Texas", 2000), ("Texas", 2010)]

populations = [33871648, 37253956,
               1897457, 19378102,
               20851828, 25145561]

pop = pd.Series(populations, index=index)
pop
```

### Pandas MultiIndex

```python
index = pd.MultiIndex.from_tuples(index)
index
```

```python
# Shows the State name once (better representation)
# Same Value is in the line above
pop = pop.reindex(index)
pop
```

```python
# Accessing all data from 2010 with Pandas slicing 
pop[:, 2010]
```

### MultiIndex as extra dimension

```python
# Data could also be stored in a DataFrame (with columns and rows labels)
# Unstacking Method, converts multi-index Series into DataFrames
pop_df = pop.unstack()
pop_df
```

```python
# Opposite: stack() method, converts DataFrame into multi-index Series
pop_df.stack()
```

```python
pop_df = pd.DataFrame({"total": pop,
"under 18": [9267089, 9284094,
             4687374, 4318033,
             5906301, 6879014]})

pop_df
```

```python
f_u18 = pop_df["under 18"] / pop_df["total"]
f_u18.unstack()
```

## Rearranging Multi-Indices

### Sorted and unsorted indicies

```python
index = pd.MultiIndex.from_product([['a', 'c', 'b'], [1, 2]])
data = pd.Series(np.random.rand(6), index=index)
data.index.names = ['char', 'int']
data
```

```python
data = data.sort_index()
data
```

```python
data["a":"b"]
```
