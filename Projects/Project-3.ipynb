```python
df["release_date"] = pd.to_datetime(df["release_date"])
df
```

```python
# Calculating ROI
df["profit"] = df["revenue"] - df["budget"]
df
```

```python
df.info()
```

```python
# Percentage Return
df["percentage_return"] = (df["profit"] / df["budget"]) * 100
df
```

```python
# Grouping 
df.groupby("genres")["profit"].mean()
```

### Phase 3 

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(np.log(df["budget"]), df["revenue"], alpha=0.5)
plt.xlabel("Budget")
plt.ylabel("Revenue")
plt.title("Does Money Buy Success?")
```

```python
genre_profit = df.groupby("genre")["profit"]

plt.figure()
genre_profit.plot(kind="bar")

plt.title("Top 10 Most Profitable Genres")
plt.xlabel("Genre")
plt.ylabel("Average Profit")

plt.show()
```

```python
plt.figure()
plt.hist(df["runtime"])
plt.title("Movie Runtime Distribution")
plt.xlabel("Runtime (minutes)")
plt.ylabel("Number of Movies")

plt.show()
```
